## What is BHJet? 

Accreting black holes can launch collimated, relativistic outflows called jets. As plasma travels away from the black hole, its density, magnetic field, speed, and particle population evolve. Those conditions determine the electromagnetic radiation that reaches us, from radio waves to high-energy gamma rays.

BHJet is a physical multi-zone model of this process. It calculates the jet dynamics, divides the flow into emitting regions, and predicts the radiation from thermal and accelerated particles in each region. Adding the regions together produces a model broadband spectral energy distribution (SED) of an extended jet, which we can compare with observations.

These tutorials begin by exploring how the physical parameters change a jet spectrum, in order to build something of an intuition of how this radiation is produced. The notebooks then build up from one emission zone to an extended multi-zone jet and compare its SED with multi-wavelength observations from a supermassive black hole (M87) and an X-ray binary (GX 339-4)
