```python
from pathlib import Path
import bhjet
import ipywidgets as widgets
from IPython.display import display

def find_repository_root(start=Path.cwd()):
    for candidate in (start.resolve(), *start.resolve().parents):
        if (candidate / 'examples' / 'gammapy' / 'data' / 'ngc4261' / 'ngc4261_dnde.csv').is_file():
            return candidate
    raise FileNotFoundError('Open this notebook from a complete modBHJet checkout.')

repository_root = find_repository_root()
print(f'BHJet: {bhjet.__file__}')
print(f'Tutorial data: {repository_root / "examples/gammapy/data/ngc4261/ngc4261_dnde.csv"}')
display(widgets.IntSlider(description='Widget check', value=5, min=0, max=10))
```
